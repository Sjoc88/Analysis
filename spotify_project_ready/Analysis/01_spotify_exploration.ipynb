```python
import sqlite3
import pandas as pd
from pathlib import Path

# Find the database
db_path = Path.cwd().parent / "spotify.db"

# Connect
conn = sqlite3.connect(db_path)

print("Database:", db_path)
print("Exists:", db_path.exists())
print("Size:", db_path.stat().st_size, "bytes")
print("Pandas:", pd.__version__)
print("SQLite:", sqlite3.sqlite_version)
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
"""

tables = pd.read_sql_query(query, conn)
tables
```

```python
query = """
PRAGMA table_info(unified_listening_history);
"""

schema = pd.read_sql_query(query, conn)
schema
```

```python
query = """
SELECT
    COUNT(*) AS events,
    COUNT(DISTINCT track_name) AS tracks,
    COUNT(DISTINCT artist_name) AS artists,
    COUNT(DISTINCT album_name) AS albums,
    ROUND(SUM(minutes_played), 1) AS minutes_played,
    ROUND(SUM(minutes_played) / 60.0, 1) AS hours_played,
    MIN(played_at) AS first_play,
    MAX(played_at) AS last_play
FROM unified_listening_history
WHERE provider = 'Spotify';
"""

spotify_overview = pd.read_sql_query(query, conn)
spotify_overview
```

```python
query = """
SELECT
    year,
    COUNT(*) AS events,
    COUNT(DISTINCT track_name) AS tracks,
    COUNT(DISTINCT artist_name) AS artists,
    COUNT(DISTINCT album_name) AS albums,
    ROUND(SUM(minutes_played), 1) AS minutes_played,
    ROUND(SUM(minutes_played) / 60.0, 1) AS hours_played
FROM unified_listening_history
WHERE provider = 'Spotify'
GROUP BY year
ORDER BY year;
"""

spotify_by_year = pd.read_sql_query(query, conn)
spotify_by_year
```
